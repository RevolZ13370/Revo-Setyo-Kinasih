#Praktikum 3: Mendesain Kelas OOP

```python
import pandas as pd
from abc import ABC, abstractmethod

class Lokasi(ABC):

    def __init__(self, nama: str, latitude: float, longitude: float):

        self.nama = str(nama) if nama else "Tanpa Nama"

        try:

            self.latitude = float(latitude)

            self.longitude = float(longitude)

        except (ValueError, TypeError, SystemError):

            self.latitude = 0.0

            self.longitude = 0.0

    def get_koordinat(self) -> tuple:
        return (self.latitude, self.longitude)

    @abstractmethod

    def get_info_popup(self) -> str:
        pass

    def __repr__(self) -> str:
        return f"{type(self).__name__}(nama='{self.nama}', lat={self.latitude:.4f}, lon={self.longitude:.4f})"

    def __str__(self) -> str:
        return f"{self.nama} [{type(self).__name__}]"

class TempatWisata(Lokasi):

    def __init__(self, nama: str, latitude: float, longitude: float, jenis: str, deskripsi: str):

        super().__init__(nama, latitude, longitude)
        self.jenis_wisata = str(jenis) if jenis else "Umum"
        self.deskripsi = str(deskripsi) if deskripsi else "Tidak ada deskripsi."

    def get_info_popup(self) -> str:
        return f"<h4><b>{self.nama}</b></h4><i>{self.jenis_wisata}</i><br><br>{self.deskripsi}<br><br>Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})"

class Kuliner(Lokasi):

    def __init__(self, nama: str, latitude: float, longitude: float, menu_andalan: str):

        super().__init__(nama, latitude, longitude)
        self.menu_andalan = str(menu_andalan) if menu_andalan else "Tidak diketahui"

    def get_info_popup(self) -> str:
        return f"<h4><b>{self.nama}</b></h4><i>Kuliner</i><br><br>Menu Andalan: {self.menu_andalan}<br><br>Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})"

class TempatIbadah(Lokasi):

    def __init__(self, nama: str, latitude: float, longitude: float, agama: str = "Umum", deskripsi: str = ""):

        super().__init__(nama, latitude, longitude)
        self.agama = str(agama) if agama else "Umum"
        self.deskripsi = str(deskripsi) if deskripsi else "Tempat Ibadah"

    def get_info_popup(self) -> str:
        return f"<h4><b>{self.nama}</b></h4><i>Tempat Ibadah ({self.agama})</i><br><br>{self.deskripsi}<br><br>Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})"



if __name__ == "__main__":
    print("--- Menguji Kelas OOP ---")


    wisata1 = TempatWisata("Lawang Sewu", -6.9840, 110.4105, "Wisata Sejarah", "Bangunan bersejarah peninggalan Belanda.")
    kuliner1 = Kuliner("Lumpia Gang Lombok", -6.9718, 110.4255, "Lumpia basah dan goreng legendaris")
    ibadah1  = TempatIbadah("Masjid Agung Jawa Tengah", -6.9892, 110.4452, "Islam", "Masjid besar dengan menara pandang Asmaul Husna.")

    print("\n--- Output __str__ ---")
    print(wisata1)
    print(kuliner1)
    print(ibadah1)

    print("\n--- Output __repr__ ---")
    print(repr(wisata1))
    print(repr(kuliner1))
    print(repr(ibadah1))

    print("\n--- get_koordinat() ---")
    print(wisata1.get_koordinat())

    print("\n--- get_info_popup() (Polimorfisme) ---")
    for obj in [wisata1, kuliner1, ibadah1]:
        print(f"\n{obj.nama}:")
        print(obj.get_info_popup())

    print("\n--- Praktikum 3 Selesai ---")
```

#Praktikum 4: Membuat Objek dari Data

```python
import pandas as pd
from abc import ABC, abstractmethod

class Lokasi(ABC):

    def __init__(self, nama: str, latitude: float, longitude: float):

        self.nama = str(nama) if nama else "Tanpa Nama"

        try:

            self.latitude = float(latitude)

            self.longitude = float(longitude)

        except (ValueError, TypeError, SystemError):

            self.latitude = 0.0

            self.longitude = 0.0

    def get_koordinat(self) -> tuple:
        return (self.latitude, self.longitude)

    @abstractmethod

    def get_info_popup(self) -> str:
        pass

    def __repr__(self) -> str:
        return f"{type(self).__name__}(nama='{self.nama}', lat={self.latitude:.4f}, lon={self.longitude:.4f})"

    def __str__(self) -> str:
        return f"{self.nama} [{type(self).__name__}]"


class TempatWisata(Lokasi):

    def __init__(self, nama: str, latitude: float, longitude: float, jenis: str, deskripsi: str):

        super().__init__(nama, latitude, longitude)
        self.jenis_wisata = str(jenis) if jenis else "Umum"
        self.deskripsi = str(deskripsi) if deskripsi else "Tidak ada deskripsi."

    def get_info_popup(self) -> str:
        return f"<h4><b>{self.nama}</b></h4><i>{self.jenis_wisata}</i><br><br>{self.deskripsi}<br><br>Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})"


class Kuliner(Lokasi):

    def __init__(self, nama: str, latitude: float, longitude: float, menu_andalan: str):

        super().__init__(nama, latitude, longitude)
        self.menu_andalan = str(menu_andalan) if menu_andalan else "Tidak diketahui"

    def get_info_popup(self) -> str:
        return f"<h4><b>{self.nama}</b></h4><i>Kuliner</i><br><br>Menu Andalan: {self.menu_andalan}<br><br>Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})"


class TempatIbadah(Lokasi):

    def __init__(self, nama: str, latitude: float, longitude: float, agama: str = "Umum", deskripsi: str = ""):

        super().__init__(nama, latitude, longitude)
        self.agama = str(agama) if agama else "Umum"
        self.deskripsi = str(deskripsi) if deskripsi else "Tempat Ibadah"

    def get_info_popup(self) -> str:
        return f"<h4><b>{self.nama}</b></h4><i>Tempat Ibadah ({self.agama})</i><br><br>{self.deskripsi}<br><br>Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})"

def baca_data_lokasi(nama_file: str):

    try:
        dataframe = pd.read_csv(nama_file)
        return dataframe

    except FileNotFoundError:
        print(f"ERROR: File '{nama_file}' tidak ditemukan!")
        return None

    except Exception as e:
        print(f"ERROR saat membaca file CSV: {type(e).__name__} - {e}")
        return None

def buat_objek_lokasi_dari_df(dataframe) -> list:

    list_objek_lokasi = []
    if dataframe is None or dataframe.empty:
        print("DataFrame kosong atau None, tidak ada objek dibuat.")
        return list_objek_lokasi

    print("\nMembuat objek dari DataFrame...")

    for index, row in dataframe.iterrows():
        nama = row.get('Nama', None)
        lat = row.get('Latitude', None)
        lon = row.get('Longitude', None)
        tipe = row.get('Tipe', 'Lainnya')
        deskripsi = row.get('Deskripsi', '')
        objek = None

        if nama is None or lat is None or lon is None:

            print(f"   -> Melewati baris {index}: Data Nama/Latitude/Longitude tidak lengkap.")

            continue
        try:

            if 'Wisata' in tipe or tipe == 'Landmark':

                objek = TempatWisata(nama, lat, lon, tipe, deskripsi)

            elif tipe == 'Kuliner':

                objek = Kuliner(nama, lat, lon, deskripsi)

            elif 'Ibadah' in tipe:

                agama_info = "Umum"  # Default

                if "Islam" in tipe: agama_info = "Islam"

                elif "Kristen" in tipe: agama_info = "Kristen"

                elif "Klenteng" in tipe: agama_info = "Tridharma"

                objek = TempatIbadah(nama, lat, lon, agama_info, deskripsi)

            else:

                print(f" -> Peringatan: Tipe '{tipe}' untuk '{nama}' tidak dikenali. Tidak membuat objek spesifik.")

            if objek:

                list_objek_lokasi.append(objek)

        except Exception as e:

            print(f" -> GAGAL membuat objek untuk '{nama}' di baris {index}: {e}")

    print(f"Total {len(list_objek_lokasi)} objek lokasi berhasil dibuat dari {len(dataframe)} baris data.")

    return list_objek_lokasi

if __name__ == "__main__":

    NAMA_FILE_CSV = "lokasi_semarang.csv"

    print("--- Memulai Praktikum 4: Membuat Objek dari Data Pandas ---")

    df_lokasi = baca_data_lokasi(NAMA_FILE_CSV)

    list_semua_lokasi = buat_objek_lokasi_dari_df(df_lokasi)

    print("\n--- Daftar Objek Lokasi yang Berhasil Dibuat ---")

    if list_semua_lokasi:

        for idx, lok in enumerate(list_semua_lokasi):

            print(f"{idx+1}. {repr(lok)}")

    else:

        print("Tidak ada objek lokasi yang dibuat.")

    print("\n--- Praktikum 4 Selesai ---")
```

#Praktikum 5: Visualisasi dengan Folium

```python
import folium
import pandas as pd

def buat_peta_lokasi_folium(list_objek: list,
                             file_output: str = "peta_lokasi.html"):
    """
    Membuat peta Folium interaktif dengan marker untuk setiap objek
    dalam list_objek.

    Args:
        list_objek  (list): List berisi instance objek turunan Lokasi.
        file_output (str) : Nama file HTML untuk menyimpan peta.
    """
    if not list_objek:
        print("Tidak ada objek lokasi untuk dipetakan.")
        return

    print(f"\nMemulai pembuatan peta Folium dari {len(list_objek)} lokasi...")

    try:
        lat_tengah = list_objek[0].latitude
        lon_tengah = list_objek[0].longitude
    except IndexError:
        lat_tengah, lon_tengah = -6.9929, 110.4200

    peta = folium.Map(
        location=[lat_tengah, lon_tengah],
        zoom_start=13,
        tiles="OpenStreetMap"
    )
    print(f"  -> Objek peta dibuat, berpusat di ({lat_tengah:.4f}, {lon_tengah:.4f})")

    jumlah_marker_valid = 0
    for lok in list_objek:
        koordinat = lok.get_koordinat()

        if koordinat != (0.0, 0.0):
            info_popup_html = lok.get_info_popup()

            folium.Marker(
                location=koordinat,
                popup=folium.Popup(info_popup_html, max_width=300),
                tooltip=lok.nama
            ).add_to(peta)

            jumlah_marker_valid += 1
        else:
            print(f"  -> Melewati marker '{lok.nama}' karena koordinat tidak valid.")

    try:
        peta.save(file_output)
        print(f"\n  -> Peta berhasil disimpan sebagai '{file_output}'.")
        print(f"     Total marker ditambahkan: {jumlah_marker_valid}")
    except Exception as e:
        print(f"\n  -> ERROR saat menyimpan peta: {type(e).__name__} - {e}")

    return peta


# Kode Utama
print("─" * 60)
print("Memulai Praktikum 5: Visualisasi Peta dengan Folium")
print("─" * 60)

NAMA_FILE_CSV  = "lokasi_semarang.csv"
NAMA_FILE_PETA = "peta_interaktif_semarang.html"

df_lokasi = baca_data_lokasi(NAMA_FILE_CSV)

list_semua_lokasi = buat_objek_lokasi_dari_df(df_lokasi)
peta = buat_peta_lokasi_folium(list_semua_lokasi, NAMA_FILE_PETA)

print(f"\nBuka file '{NAMA_FILE_PETA}' di browser untuk melihat hasilnya.")
print("\n─── Tampilan Peta Langsung di Notebook ───")
peta
```

#Praktikum 6: File Handling Tambahan (Log)

```python
import pandas as pd
import folium
import datetime

def tulis_log(pesan: str, file_log: str = "proses_peta.log"):
    """
    Menulis pesan log ke file dengan timestamp menggunakan mode append ('a').

    Args:
        pesan    (str): Pesan yang akan dicatat.
        file_log (str): Nama file log tujuan.
    """
    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    try:
        with open(file_log, 'a', encoding='utf-8') as f:
            f.write(f"[{timestamp}] {pesan}\n")
    except IOError as e:
        print(f"ERROR: Gagal menulis ke file log '{file_log}': {e}")

def buat_peta_lokasi_folium_dengan_log(list_objek: list,
                                        file_output: str = "peta_lokasi.html",
                                        file_log: str = "proses_peta.log"):
    """
    Membuat peta Folium, menambahkan marker, menyimpan ke HTML,
    dan menulis log proses ke file.
    """
    nama_fungsi = "buat_peta_lokasi_folium"

    if not list_objek:
        pesan_log = f"[{nama_fungsi}] GAGAL: Tidak ada data lokasi untuk dipetakan."
        print(pesan_log)
        tulis_log(pesan_log, file_log)
        return None

    pesan_mulai = f"[{nama_fungsi}] Memulai pembuatan peta '{file_output}' dengan {len(list_objek)} lokasi."
    print(f"\n{pesan_mulai}")
    tulis_log(pesan_mulai, file_log)

    try:
        lat_tengah = list_objek[0].latitude
        lon_tengah = list_objek[0].longitude
    except IndexError:
        lat_tengah, lon_tengah = -6.9929, 110.4200

    peta = folium.Map(location=[lat_tengah, lon_tengah], zoom_start=12)

    jumlah_marker  = 0
    lokasi_dilewati = []

    for lok in list_objek:
        koordinat = lok.get_koordinat()
        if koordinat != (0.0, 0.0):
            info_popup_html = lok.get_info_popup()
            folium.Marker(
                location=koordinat,
                popup=folium.Popup(info_popup_html, max_width=300),
                tooltip=lok.nama
            ).add_to(peta)
            jumlah_marker += 1
        else:
            lokasi_dilewati.append(lok.nama)

    if lokasi_dilewati:
        pesan_lewat = (f"[{nama_fungsi}] Melewati marker untuk: "
                       f"{', '.join(lokasi_dilewati)} (koordinat tidak valid).")
        print(f"  -> Peringatan: {pesan_lewat}")
        tulis_log(pesan_lewat, file_log)

    try:
        peta.save(file_output)
        pesan_sukses = (f"[{nama_fungsi}] Peta '{file_output}' berhasil dibuat "
                        f"dengan {jumlah_marker} marker.")
        print(f"  -> {pesan_sukses}")
        tulis_log(pesan_sukses, file_log)
    except Exception as e:
        pesan_error = (f"[{nama_fungsi}] ERROR saat menyimpan peta "
                       f"'{file_output}': {type(e).__name__} - {e}")
        print(f"  -> {pesan_error}")
        tulis_log(pesan_error, file_log)

    return peta


# Kode Utama
print("─" * 60)
print("Memulai Praktikum 6: File Handling Tambahan (Log)")
print("─" * 60)

NAMA_FILE_CSV  = "lokasi_semarang.csv"
NAMA_FILE_PETA = "peta_interaktif_semarang.html"
FILE_LOG       = "proses_peta.log"

df_lokasi = baca_data_lokasi(NAMA_FILE_CSV)

list_semua_lokasi = buat_objek_lokasi_dari_df(df_lokasi)

peta6 = buat_peta_lokasi_folium_dengan_log(list_semua_lokasi, NAMA_FILE_PETA, FILE_LOG)

print("\nMenjalankan pembuatan peta kedua (demo log append)...")
buat_peta_lokasi_folium_dengan_log(list_semua_lokasi, "peta_kedua.html", FILE_LOG)

print(f"\n─── Isi File Log '{FILE_LOG}' ───")
try:
    with open(FILE_LOG, 'r', encoding='utf-8') as f:
        isi_log = f.readlines()
    for baris_log in isi_log:
        print(f"  {baris_log}", end='')
except FileNotFoundError:
    print(f"  File log '{FILE_LOG}' belum ada.")

print("\n\n─── Praktikum 6 Selesai ───")
```

```python
import pandas as pd
import folium
import datetime
from abc import ABC, abstractmethod

# ─────────────────────────────────────────────────────────────
# (a) TAMBAH DATA & KELAS BARU
# ─────────────────────────────────────────────────────────────

# ── Update CSV dengan 3 lokasi baru ──────────────────────────
csv_content_extended = """Nama,Latitude,Longitude,Tipe,Deskripsi
Lawang Sewu,-6.9840,110.4105,Wisata Sejarah,Bangunan bersejarah peninggalan Belanda dengan banyak pintu.
Simpang Lima,-6.9929,110.4200,Landmark,"Alun-alun pusat kota Semarang, tempat berkumpul populer."
Masjid Agung Jawa Tengah,-6.9892,110.4452,Tempat Ibadah,Masjid besar dengan arsitektur megah dan menara pandang Asmaul Husna.
Klenteng Sam Poo Kong,-6.9980,110.4030,Tempat Ibadah,Klenteng bersejarah peninggalan Laksamana Cheng Ho.
Brown Canyon,-7.0375,110.4875,Wisata Alam,"Tebing bekas penambangan galian C yang unik menyerupai Grand Canyon."
Lumpia Gang Lombok,-6.9718,110.4255,Kuliner,Salah satu tempat makan lumpia legendaris dan otentik di Semarang.
Kota Lama,-6.9690,110.4250,Wisata Sejarah,"Kawasan cagar budaya dengan bangunan-bangunan kuno bergaya Eropa."
Pantai Marina,-6.9585,110.3875,Wisata Alam,Kawasan pantai rekreasi di bagian utara kota.
Kampoeng Kopi Banaran,-7.2780,110.4010,Wisata Alam,"Agrowisata kebun kopi dengan pemandangan dan restoran."
Toko Oen,-6.9715,110.4235,Kuliner,Restoran dan toko es krim legendaris sejak zaman kolonial.
Balai Kota Semarang,-6.9876,110.4206,Kantor Pemerintahan,Kantor wali kota Semarang sebagai pusat pemerintahan kota.
Museum Jawa Tengah Ranggawarsita,-6.9971,110.4067,Museum,"Museum terbesar di Jawa Tengah yang menyimpan koleksi sejarah, budaya, dan alam."
Taman Indonesia Kaya,-6.9819,110.4113,Taman Kota,"Taman terbuka hijau di pusat kota dengan area pertunjukan seni budaya."
"""

NAMA_FILE_CSV_EXT = "lokasi_semarang_extended.csv"
with open(NAMA_FILE_CSV_EXT, 'w', encoding='utf-8') as f:
    f.write(csv_content_extended)

print(f"[OK] File '{NAMA_FILE_CSV_EXT}' berhasil dibuat dengan data yang diperluas.")
print(f"     Total baris data: 13 lokasi (10 asli + 3 baru: Kantor, Museum, Taman)\n")

# ── Definisi ulang kelas lama ──────────────────────────────────
class Lokasi(ABC):
    def __init__(self, nama, latitude, longitude):
        self.nama = str(nama) if nama else "Tanpa Nama"
        try:
            self.latitude  = float(latitude)
            self.longitude = float(longitude)
        except (ValueError, TypeError):
            self.latitude = self.longitude = 0.0

    def get_koordinat(self): return (self.latitude, self.longitude)

    @abstractmethod
    def get_info_popup(self) -> str: pass

    def __repr__(self):
        return f"{type(self).__name__}(nama='{self.nama}', lat={self.latitude:.4f}, lon={self.longitude:.4f})"
    def __str__(self): return f"{self.nama} [{type(self).__name__}]"


class TempatWisata(Lokasi):
    def __init__(self, nama, latitude, longitude, jenis, deskripsi):
        super().__init__(nama, latitude, longitude)
        self.jenis_wisata = str(jenis) if jenis else "Umum"
        self.deskripsi    = str(deskripsi) if deskripsi else "Tidak ada deskripsi."
    def get_info_popup(self):
        return (f"<h4><b>{self.nama}</b></h4><i>{self.jenis_wisata}</i><br><br>"
                f"{self.deskripsi}<br><br>Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


class Kuliner(Lokasi):
    def __init__(self, nama, latitude, longitude, menu_andalan):
        super().__init__(nama, latitude, longitude)
        self.menu_andalan = str(menu_andalan) if menu_andalan else "Tidak diketahui"
    def get_info_popup(self):
        return (f"<h4><b>{self.nama}</b></h4><i>Kuliner</i><br><br>"
                f"Menu Andalan: {self.menu_andalan}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


class TempatIbadah(Lokasi):
    def __init__(self, nama, latitude, longitude, agama="Umum", deskripsi=""):
        super().__init__(nama, latitude, longitude)
        self.agama     = str(agama)     if agama     else "Umum"
        self.deskripsi = str(deskripsi) if deskripsi else "Tempat Ibadah"
    def get_info_popup(self):
        return (f"<h4><b>{self.nama}</b></h4><i>Tempat Ibadah ({self.agama})</i><br><br>"
                f"{self.deskripsi}<br><br>Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


# ── KELAS BARU 1: Kantor ───────────────────────────────────────
class Kantor(Lokasi):
    """Representasi kantor pemerintahan atau instansi resmi."""
    def __init__(self, nama, latitude, longitude, instansi, deskripsi=""):
        super().__init__(nama, latitude, longitude)
        self.instansi  = str(instansi) if instansi else "Pemerintahan"
        self.deskripsi = str(deskripsi) if deskripsi else "Kantor instansi resmi."
    def get_info_popup(self):
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>Kantor Pemerintahan</i><br><br>"
                f"Instansi: {self.instansi}<br><br>"
                f"{self.deskripsi}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


# ── KELAS BARU 2: Museum ──────────────────────────────────────
class Museum(Lokasi):
    """Representasi museum atau galeri seni."""
    def __init__(self, nama, latitude, longitude, koleksi_utama, deskripsi=""):
        super().__init__(nama, latitude, longitude)
        self.koleksi_utama = str(koleksi_utama) if koleksi_utama else "Umum"
        self.deskripsi     = str(deskripsi) if deskripsi else "Museum koleksi budaya."
    def get_info_popup(self):
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>Museum</i><br><br>"
                f"Koleksi Utama: {self.koleksi_utama}<br><br>"
                f"{self.deskripsi}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


# ── KELAS BARU 3: TamanKota ───────────────────────────────────
class TamanKota(Lokasi):
    """Representasi taman kota / ruang terbuka hijau."""
    def __init__(self, nama, latitude, longitude, fasilitas, deskripsi=""):
        super().__init__(nama, latitude, longitude)
        self.fasilitas = str(fasilitas) if fasilitas else "Taman Umum"
        self.deskripsi = str(deskripsi) if deskripsi else "Ruang terbuka hijau."
    def get_info_popup(self):
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>Taman Kota</i><br><br>"
                f"Fasilitas: {self.fasilitas}<br><br>"
                f"{self.deskripsi}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


print("[OK] Semua kelas berhasil didefinisikan:")
print("     Lokasi (abstrak) → TempatWisata, Kuliner, TempatIbadah, Kantor, Museum, TamanKota")
```

```python
# ─────────────────────────────────────────────────────────────
# (a) FUNGSI BUAT OBJEK (DIPERBARUI – kenali tipe baru)
# ─────────────────────────────────────────────────────────────

def baca_data_lokasi(nama_file):
    try:
        df = pd.read_csv(nama_file)
        print(f"[OK] File '{nama_file}' berhasil dibaca. ({len(df)} baris)")
        return df
    except FileNotFoundError:
        print(f"[ERROR] File '{nama_file}' tidak ditemukan!")
        return None
    except Exception as e:
        print(f"[ERROR] {type(e).__name__}: {e}")
        return None


def buat_objek_lokasi_dari_df(dataframe):
    list_objek = []
    if dataframe is None or dataframe.empty:
        return list_objek

    for index, row in dataframe.iterrows():
        nama      = row.get('Nama',      None)
        lat       = row.get('Latitude',  None)
        lon       = row.get('Longitude', None)
        tipe      = str(row.get('Tipe', 'Lainnya'))
        deskripsi = row.get('Deskripsi', '')

        if nama is None or lat is None or lon is None:
            continue

        objek = None
        try:
            if 'Wisata' in tipe or tipe == 'Landmark':
                objek = TempatWisata(nama, lat, lon, tipe, deskripsi)

            elif tipe == 'Kuliner':
                objek = Kuliner(nama, lat, lon, deskripsi)

            elif 'Ibadah' in tipe:
                agama = "Umum"
                if "Masjid" in nama or "Islam" in tipe: agama = "Islam"
                elif "Gereja" in nama or "Kristen" in tipe: agama = "Kristen"
                elif "Klenteng" in nama: agama = "Tridharma"
                objek = TempatIbadah(nama, lat, lon, agama, deskripsi)

            # ── Tipe Baru ──────────────────────────────────────
            elif 'Kantor' in tipe or 'Pemerintahan' in tipe:
                objek = Kantor(nama, lat, lon, tipe, deskripsi)

            elif 'Museum' in tipe:
                objek = Museum(nama, lat, lon, "Sejarah & Budaya", deskripsi)

            elif 'Taman' in tipe:
                objek = TamanKota(nama, lat, lon, "Taman Umum & Pertunjukan", deskripsi)
            # ───────────────────────────────────────────────────

            else:
                print(f"  [Peringatan] Tipe '{tipe}' untuk '{nama}' tidak dikenali.")

            if objek:
                list_objek.append(objek)

        except Exception as e:
            print(f"  [GAGAL] '{nama}': {e}")

    print(f"[OK] Total {len(list_objek)} objek berhasil dibuat dari {len(dataframe)} baris data.")
    return list_objek
```

```python
# ─────────────────────────────────────────────────────────────
# (b) KUSTOMISASI MARKER FOLIUM
# ─────────────────────────────────────────────────────────────

def tentukan_ikon_marker(lok):
    """
    Menentukan warna dan ikon marker Folium berdasarkan tipe objek lokasi.
    Menggunakan isinstance() untuk memeriksa tipe objek.

    Returns:
        dict: {'color': warna_str, 'icon': nama_ikon_str}
    """
    if isinstance(lok, TempatWisata):
        return {'color': 'blue',   'icon': 'camera'}      # Biru  – Wisata
    elif isinstance(lok, Kuliner):
        return {'color': 'red',    'icon': 'cutlery'}     # Merah – Kuliner
    elif isinstance(lok, TempatIbadah):
        return {'color': 'purple', 'icon': 'star'}        # Ungu  – Ibadah
    elif isinstance(lok, Kantor):
        return {'color': 'gray',   'icon': 'briefcase'}   # Abu   – Kantor
    elif isinstance(lok, Museum):
        return {'color': 'orange', 'icon': 'book'}        # Oranye – Museum
    elif isinstance(lok, TamanKota):
        return {'color': 'green',  'icon': 'leaf'}        # Hijau – Taman
    else:
        return {'color': 'black',  'icon': 'info-sign'}   # Default


# ─────────────────────────────────────────────────────────────
# (c) BACA KONFIGURASI PETA DARI FILE
# ─────────────────────────────────────────────────────────────

CONFIG_FILE = "config_peta.txt"

# Buat file konfigurasi
with open(CONFIG_FILE, 'w', encoding='utf-8') as f:
    f.write("-6.9929\n")   # latitude tengah
    f.write("110.4200\n")  # longitude tengah
    f.write("13\n")        # zoom_start

print(f"[OK] File konfigurasi '{CONFIG_FILE}' berhasil dibuat.")

def baca_konfigurasi_peta(file_config: str = "config_peta.txt") -> dict:
    """
    Membaca konfigurasi peta dari file teks (mode 'r').
    Format file: baris 1 = latitude, baris 2 = longitude, baris 3 = zoom_start.

    Returns:
        dict: {'lat': float, 'lon': float, 'zoom': int}
    """
    # Nilai default (koordinat pusat Semarang)
    config = {'lat': -6.9929, 'lon': 110.4200, 'zoom': 13}

    try:
        with open(file_config, 'r', encoding='utf-8') as f:
            baris_config = f.readlines()

        config['lat']  = float(baris_config[0].strip())
        config['lon']  = float(baris_config[1].strip())
        config['zoom'] = int(baris_config[2].strip())

        print(f"[OK] Konfigurasi peta berhasil dibaca dari '{file_config}':")
        print(f"     Latitude : {config['lat']}")
        print(f"     Longitude: {config['lon']}")
        print(f"     Zoom     : {config['zoom']}")

    except FileNotFoundError:
        print(f"[Peringatan] File konfigurasi '{file_config}' tidak ditemukan. Menggunakan nilai default.")
    except (ValueError, IndexError) as e:
        print(f"[Peringatan] Error membaca konfigurasi: {e}. Menggunakan nilai default.")

    return config
```

```python
# ─────────────────────────────────────────────────────────────
# FUNGSI LOG
# ─────────────────────────────────────────────────────────────

def tulis_log(pesan: str, file_log: str = "proses_peta.log"):
    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    try:
        with open(file_log, 'a', encoding='utf-8') as f:
            f.write(f"[{timestamp}] {pesan}\n")
    except IOError as e:
        print(f"ERROR menulis log: {e}")


# ─────────────────────────────────────────────────────────────
# (d) FUNGSI BUAT PETA – INTEGRASI PENUH
# ─────────────────────────────────────────────────────────────

def buat_peta_mini_project(list_objek: list,
                            file_output: str = "peta_mini_project.html",
                            file_config: str = "config_peta.txt",
                            file_log: str = "proses_peta.log"):
    """
    Fungsi utama mini project SIG:
    - Membaca konfigurasi dari file teks (mode 'r')
    - Membuat peta Folium dengan marker berkustomisasi per tipe
    - Menyimpan peta ke HTML
    - Mencatat log proses ke file (mode 'a')
    """
    if not list_objek:
        tulis_log("GAGAL: Tidak ada objek untuk dipetakan.", file_log)
        print("[ERROR] List objek kosong.")
        return None

    # (c) Baca konfigurasi dari file
    config = baca_konfigurasi_peta(file_config)

    tulis_log(f"Memulai pembuatan peta '{file_output}' dengan {len(list_objek)} lokasi.", file_log)

    # Buat peta Folium menggunakan nilai dari config
    peta = folium.Map(
        location=[config['lat'], config['lon']],
        zoom_start=config['zoom'],
        tiles="OpenStreetMap"
    )

    # Tambahkan legenda warna sebagai HTML overlay
    legenda_html = """
    <div style="position: fixed; bottom: 30px; left: 30px; z-index: 1000;
                background-color: white; padding: 12px; border-radius: 8px;
                border: 2px solid grey; font-size: 13px;">
      <b>Legenda</b><br>
      <i class="fa fa-map-marker" style="color:blue"></i>   Wisata &amp; Landmark<br>
      <i class="fa fa-map-marker" style="color:red"></i>    Kuliner<br>
      <i class="fa fa-map-marker" style="color:purple"></i> Tempat Ibadah<br>
      <i class="fa fa-map-marker" style="color:gray"></i>   Kantor Pemerintahan<br>
      <i class="fa fa-map-marker" style="color:orange"></i> Museum<br>
      <i class="fa fa-map-marker" style="color:green"></i>  Taman Kota<br>
    </div>
    """
    peta.get_root().html.add_child(folium.Element(legenda_html))

    jumlah_marker  = 0
    lokasi_dilewati = []

    for lok in list_objek:
        koordinat = lok.get_koordinat()

        if koordinat == (0.0, 0.0):
            lokasi_dilewati.append(lok.nama)
            continue

        # (b) Tentukan ikon & warna berdasarkan tipe objek
        ikon_info = tentukan_ikon_marker(lok)
        info_popup = lok.get_info_popup()     # Polimorfisme

        folium.Marker(
            location=koordinat,
            popup=folium.Popup(info_popup, max_width=300),
            tooltip=f"[{type(lok).__name__}] {lok.nama}",
            icon=folium.Icon(
                color=ikon_info['color'],
                icon=ikon_info['icon'],
                prefix='fa'
            )
        ).add_to(peta)
        jumlah_marker += 1

    if lokasi_dilewati:
        tulis_log(f"Melewati: {', '.join(lokasi_dilewati)} (koordinat invalid).", file_log)

    # Simpan peta
    try:
        peta.save(file_output)
        pesan_sukses = f"Peta '{file_output}' berhasil disimpan. Marker: {jumlah_marker}."
        print(f"[OK] {pesan_sukses}")
        tulis_log(pesan_sukses, file_log)
    except Exception as e:
        pesan_err = f"ERROR menyimpan peta: {type(e).__name__} - {e}"
        print(f"[ERROR] {pesan_err}")
        tulis_log(pesan_err, file_log)

    return peta
```

```python
# ─────────────────────────────────────────────────────────────
# KODE UTAMA MINI PROJECT SIG
# ─────────────────────────────────────────────────────────────

print("=" * 60)
print("  MINI PROJECT SIG – SEMARANG LOCATION MAP")
print("=" * 60)

NAMA_FILE_CSV_EXT  = "lokasi_semarang_extended.csv"
NAMA_FILE_PETA_MP  = "peta_mini_project.html"
CONFIG_FILE        = "config_peta.txt"
FILE_LOG           = "proses_peta.log"

# 1. Baca data CSV yang sudah diperluas
print("\n[STEP 1] Membaca data CSV...")
df_ext = baca_data_lokasi(NAMA_FILE_CSV_EXT)

# 2. Buat list objek (termasuk tipe baru)
print("\n[STEP 2] Membuat objek dari DataFrame...")
list_objek_ext = buat_objek_lokasi_dari_df(df_ext)

# 3. Tampilkan ringkasan objek berdasarkan tipe
from collections import Counter
tipe_count = Counter(type(lok).__name__ for lok in list_objek_ext)
print("\n─── Ringkasan Objek per Tipe ───")
for tipe_cls, jml in sorted(tipe_count.items()):
    print(f"  {tipe_cls:<20s}: {jml} lokasi")

# 4. Buat peta mini project (dengan konfigurasi, kustomisasi marker, dan logging)
print("\n[STEP 3] Membuat peta interaktif...")
peta_mp = buat_peta_mini_project(
    list_objek_ext,
    file_output=NAMA_FILE_PETA_MP,
    file_config=CONFIG_FILE,
    file_log=FILE_LOG
)

# 5. Tampilkan isi log
print("\n[STEP 4] Isi File Log:")
print("─" * 60)
with open(FILE_LOG, 'r', encoding='utf-8') as f:
    for line in f:
        print(f"  {line}", end='')

print("\n" + "─" * 60)
print(f"Buka '{NAMA_FILE_PETA_MP}' di browser untuk melihat peta interaktif.")
print("=" * 60)
print("  MINI PROJECT SELESAI")
print("=" * 60)
```
