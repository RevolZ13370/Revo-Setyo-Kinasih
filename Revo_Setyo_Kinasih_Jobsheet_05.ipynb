# Penugasan 05. Operasi dalam F-String

```python
a = int(input("Masukkan bilangan pertama : "))
b = int(input("Masukkan bilangan kedua : "))

print(f"Penjumlahan: {a} + {b} = {a + b}")
print(f"Perkalian   : {a} * {b} = {a * b}")
print(f"Pembagian   : {a} / {b} = {a / b:.3f}")
```

# Penugasan 04. Casting Data

```python
angka = str(input("Masukkan angka : "))

kuadrat = pow(int(angka), 2)
desimal = float(angka)

print(f"Kuadrat dari {angka} adalah {kuadrat}")
print(f"Desimal dari {angka} adalah {desimal:.2f}")
```

# Penugasan 05. Studi Kasus – Analisis Teks

```python
kalimat = input("Masukkan kalimat : ")
daftar_kata = kalimat.split()

# Hitung jumlah kata
print(f"Jumlah kata: {len(daftar_kata)}")

# Cari kata terpanjang
kata_terpanjang = max(daftar_kata, key=len)
print(f"Kata terpanjang: {kata_terpanjang}")

# Tampilkan daftar kata urut abjad
print("Daftar kata (abjad):")
for kata in sorted(daftar_kata):
    print(kata)
```
