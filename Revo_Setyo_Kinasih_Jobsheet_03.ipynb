# Praktikum 01. If Sederhana

```python
nilai = int(input("Masukkan nilai ujian: "))

if nilai >= 60:
  print("Anda LULUS")
```

# Praktikum 02. If-Else

```python
nilai = int(input("Masukkan nilai ujian: "))

if nilai >= 60:
  print("Anda LULUS")

else:
  print("Anda TIDAK LULUS")
```

# Praktikum 03. If-Elif-Else

```python
nilai = int(input("Masukkan nilai akhir: "))

if nilai >= 85:
  grade = "A"
elif nilai >= 70:
  grade = "B"
elif nilai >= 55:
  grade = "C"
elif nilai >= 40:
  grade = "D"
else:
  grade = "E"

print(f"Nilai Anda {nilai}, maka grade Anda = {grade}")
```

# Praktikum 04. Nested If

```python
ipk = float(input("Masukkan IPK: "))
sks = int(input("Masukkan jumlah SKS: "))

if ipk >= 3.0:

  if sks >= 18:
    print("Anda LAYAK mendapat beasiswa")
  else:
    print("IPK memenuhi, tapi SKS kurang")

else:
  print("Anda TIDAK layak mendapat beasiswa")
```

# Praktikum 05. Match-Case

```python
menu = int(input("Pilih menu (1-3): "))

match menu:
  case 1:
    print("Anda memilih Nasi Goreng")
  case 2:
    print("Anda memilih Mie Ayam")
  case 3:
    print("Anda memilih Soto Ayam")
  case _:
    print("Menu tidak tersedia")
```

# Praktikum 06. Studi Kasus: Program Kasir

```python
barang = input("Masukkan nama barang: ")
harga = float(input("Masukkan harga barang: "))
jumlah = int(input("Masukkan jumlah barang: "))

total = harga * jumlah

if total >= 500000:
  diskon = 0.2
elif total >= 250000:
  diskon = 0.1
else:
  diskon = 0

bayar = total - (total * diskon)

print("\n===== Struk Belanja =====")
print(f"Barang : {barang}")
print(f"Jumlah : {jumlah}")
print(f"Total : Rp {total}")
print(f"Diskon : {diskon*100}%")
print(f"Bayar : Rp {bayar}")
```

# Penugasan 01. Cek Bilangan Positif/Negatif/Nol

```python
# Input bilangan bulat
bilangan = int(input("Masukkan bilangan bulat: "))

# Output untuk menentukan bilang positif/negatif/nol
if bilangan > 0:
  print("Bilangan positif")
elif bilangan < 0:
  print("Bilangan negatif")
else:
  print("Bilangan nol")
```

# Penugasan 02. Menentukan Kategori Usia

```python
# Input usia menggunakan integer
usia = int(input("Masukkan usia: "))

# Output kategori berdasarkan usia yang sudah diinput
if usia >= 0 and usia < 5:
  print("Anda adalah Balita")
elif usia >= 5 and usia <= 11:
  print("Anda adalah Anak-anak")
elif usia >= 12 and usia <= 17:
  print("Anda adalah Remaja")
elif usia >= 18 and usia <= 59:
  print("Anda adalah Dewasa")
else:
  print("Anda adalah Lansia")
```

# Penugasan 03. Hitung Tarif Parkir

```python
# Input durasi parkir
durasi = int(input("Masukkan durasi parkir (jam): "))

# Set tarif awal
tarif = 5000

# Jika parkir lebih dari 2 jam, hitung tarif tambahan
if durasi > 2:
  tarif += (durasi - 2) * 2000

# Output tarif parkir
print(f"Tarif parkir Anda adalah Rp {tarif}")
```

# Penugasan 06. Menu Restoran dengan Match-Case

```python
# Input menu yang akan dipilih
menu = int(input("Pilih menu (1-3): "))

# Pilihan menu yang dapat dipilih
match menu:
  case 1:
    print("Anda memesan Nasi Goreng")
  case 2:
    print("Anda memesan Mie Ayam")
  case 3:
    print("Anda memesan Soto Ayam")
  case _:
    print("Menu tidak tersedia")
```

# Penugasan 07. Studi Kasus Nyata – Program Penentuan Diskon

```python
# Input barang, harga, dan jumlah
barang = input("Masukkan nama barang: ")
harga = float(input("Masukkan harga barang: "))
jumlah = int(input("Masukkan jumlah barang: "))

# Logika penjumlahan total harga
total = harga * jumlah

# Logika diskon menggunakan if-elif-else
if total >= 500000:
  diskon = 0.2
elif total >= 250000:
  diskon = 0.1
else:
  diskon = 0

# Logika harga yang harus dibayar
bayar = total - (total * diskon)

# Output Struk Belanja
print("\n===== Struk Belanja =====")
print(f"Barang : {barang}")
print(f"Jumlah : {jumlah}")
print(f"Total : Rp {total}")
print(f"Diskon : {diskon*100}%")
print(f"Bayar : Rp {bayar}")
```
