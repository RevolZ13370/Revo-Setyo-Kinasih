# Praktikum 01 - Tipe data dasar

```python
angka_int = 10
print("angka_int =", angka_int, "->", type(angka_int))

# Float
angka_float = 3.14
print("angka_float =", angka_float, "->", type(angka_float))

# Complex
angka_complex = 2 + 3j
print("angka_complex =", angka_complex, "->", type(angka_complex))

# Boolean
is_active = True
print("is_active =", is_active, "->", type(is_active))

# String
teks = "Hello, Python!"
print("teks =", teks, "->", type(teks))

# List
daftar_angka = [1, 2, 3, 4, 5]
print("daftar_angka =", daftar_angka, "->", type(daftar_angka))

# Tuple
koordinat = (10, 20)
print("koordinat =", koordinat, "->", type(koordinat))

# Dictionary
data_mahasiswa = {
  "nama": "Andi",
  "nim": "A11.2022.12345",
  "jurusan": "Teknik Informatika"
  }
print("data_mahasiswa =", data_mahasiswa, "->", type(data_mahasiswa))

# Set
himpunan_angka = {1, 2, 3, 2, 1}
print("himpunan_angka =", himpunan_angka, "->", type(himpunan_angka))

# Contoh penggunaan konversi tipe data
nilai_str = "100"
print("\nnilai_str =", nilai_str, "->", type(nilai_str))
nilai_int = int(nilai_str)
print("nilai_int =", nilai_int, "->", type(nilai_int))
```

# Praktikum 02 - Variable dan operasi dasar

```python
# Pendeklarasian Variabel
nama = "Budi"
umur = 20
tinggi = 170.5
is_student = True

print("Nama =", nama)
print("Umur =", umur)
print("Tinggi =", tinggi, "cm")
print("Mahasiswa =", is_student)

# Operasi Aritmetika
a = 10
b = 3

penjumlahan = a + b
pengurangan = a - b
perkalian = a * b
pembagian = a / b
pembagian_bulat = a // b
modulus = a % b
pangkat = a ** b

print("\nOPERASI ARITMETIKA")
print("a =", a, ", b =", b)
print("Penjumlahan =", penjumlahan)
print("Pengurangan =", pengurangan)
print("Perkalian =", perkalian)
print("Pembagian =", pembagian)
print("Pembagian Bulat =", pembagian_bulat)
print("Modulus =", modulus)
print("Pangkat =", pangkat)

# Operasi Perbandingan
print("\nOPERASI PERBANDINGAN")
print("a > b =", a > b)
print("a < b =", a < b)
print("a == b =", a == b)
print("a != b =", a != b)
print("a >= b =", a >= b)
print("a <= b =", a <= b)

# Operasi Logika
x = True
y = False

print("\nOPERASI LOGIKA")
print("x =", x, ", y =", y)
print("x and y =", x and y)
print("x or y =", x or y)
print("not x =", not x)
```

# Praktikum 03 - Percabangan

```python
# IF Sederhana
nilai = 85
if nilai > 80:
    print("Selamat! Anda lulus dengan nilai tinggi.\n")

# IF-ELSE
umur = 17
if umur >= 18:
    print("Anda sudah cukup umur untuk mendapatkan SIM.")
else:
    print("Anda belum cukup umur untuk mendapatkan SIM.\n")

# IF-ELIF-ELSE
hari = "Rabu"
if hari == "Senin":
    print("Hari Senin - Saatnya kembali bekerja!")
elif hari == "Selasa":
    print("Hari Selasa - Jadwal rapat mingguan.")
elif hari == "Rabu":
    print("Hari Rabu - Ada diskon di beberapa toko.")
else:
    print("Hari lainnya - Atur jadwalmu dengan baik.\n")

# IF Bersarang (Nested IF)
suhu = 35
if suhu > 30:
    print("Cuaca cukup panas.")
    if suhu > 40:
        print("Bahkan sangat terik! Disarankan banyak minum air.")
    else:
        print("Masih relatif normal, tapi tetap jaga kesehatan.")
else:
    print("Cuaca sepertinya cukup sejuk.\n")

# Operasi Logika dalam Percabangan
nilai_teori = 75
nilai_praktik = 80
if nilai_teori >= 70 and nilai_praktik >= 70:
    print("Anda lulus karena nilai teori dan praktik memadai.")

# If Ternary
angka = -5
status = "Positif" if angka > 0 else "Negatif atau Nol"
print(f"Angka = {angka} => {status}")
```

# Praktikum 04 - Perulangan

```python
# FOR Loop dengan range()
print("1) FOR loop dengan range()")
for i in range(5):
    print("Perulangan ke-", i)

# FOR Loop mengiterasi List
buah = ["apel", "mangga", "jeruk", "pisang"]
for item in buah:
    print("Buah:", item)

# WHILE Loop
count = 0
while count < 5:
    print("count =", count)
    count += 1

# BREAK dan CONTINUE
print("\n4 & 5) BREAK dan CONTINUE")
for i in range(5):
    if i == 2:
        continue # Lewati iterasi i=2
    if i == 4:
        break    # Hentikan loop saat i=4
    print("i =", i)

# NESTED Loop [cite: 314]
print("\n6) NESTED loop")
for i in range(3):
    for j in range(2):
        print(f"i={i}, j={j}")

# ELSE pada Loop
for x in range(3):
    print("x =", x)
else:
    print("Loop for telah selesai tanpa break.\n")

# PASS (Placeholder)
for i in range(3):
    if i == 1:
        pass # Belum ada logika, gunakan pass agar tidak error
    print("i =", i)
```

# Buat Program 1: Buat program untuk menghitung BMI (Body Mass Index)

```python
print("--- Program Hitung BMI ---")
berat = float(input("Masukkan berat badan (kg): "))
tinggi_cm = float(input("Masukkan tinggi badan (cm): "))

# Konversi tinggi ke meter
tinggi_m = tinggi_cm / 100
bmi = berat / (tinggi_m ** 2)

print(f"Nilai BMI Anda: {bmi:.2f}")

if bmi < 18.5:
    kategori = "Underweight"
elif 18.5 <= bmi <= 24.9:
    kategori = "Normal"
else:
    kategori = "Overweight"

print(f"Kategori: {kategori}")
```

# Buat Program 2: Buat program yang mengecek apakah suatu bilangan genap, ganjil, atau prima.

```python
print("--- Program Cek Bilangan ---")
angka = int(input("Masukkan sebuah bilangan bulat positif: "))

# Cek Genap atau Ganjil
if angka % 2 == 0:
    print(f"{angka} adalah bilangan Genap")
else:
    print(f"{angka} adalah bilangan Ganjil")

# Cek Bilangan Prima
is_prima = True
if angka <= 1:
    is_prima = False
else:
    for i in range(2, int(angka**0.5) + 1):
        if angka % i == 0:
            is_prima = False
            break

if is_prima:
    print(f"{angka} adalah bilangan Prima")
else:
    print(f"{angka} bukan bilangan Prima")
```
