# Praktikum 01. Perulangan FOR

```python
for i in range (1, 6):
  print(i)
```

# Praktikum 02. Perulangan WHILE

```python
# Input nilai n
n = int(input("Masukkan angka : "))

# Logika pengulangan menggunakan while
while n > 0:
  print (n)
  n -= 1
```

# Praktikum 03. Nested Loop

```python
for i in range(1, 6):
  for j in range(i):
    print('*', end=" ")
  print(" ")
```

# Praktikum 04. Break and continue

# Break

```python
for i in range(1, 11):
  print(i)
  if i == 5:
    print("Angka 5 ditemukan")
    break
```

# Continue

```python
for i in range(1, 6):
    if i % 2 == 1:
        continue
    print(i)
```

# Praktikum 05. For pada Dictionary (Key, Value)

```python
data = {"Nama": "Revo", "Umur": 18, "Prodi": "TRK"}

for k, v in data.items():
  print(k, ":", v)
```

# Praktikum 06. For dengan Enumerate

```python
daftar = ["Andi", "Budi", "Cici"]
for i, nama in enumerate(daftar, start=1):
  print(i, nama)
```

# Praktikum 07. Studi Kasus: Program Kasir Multi-Item

```python
total = 0

while True:
  barang = input("Masukkan nama barang (ketik 'selesai' untuk berhenti): ")
  if barang.lower() == "selesai":
    break
  harga = float(input(f"Masukkan harga {barang}: "))
  jumlah = int(input(f"Masukkan jumlah {barang}: "))
  subtotal = harga * jumlah
  total += subtotal

print("\n===== Struk Belanja =====")
print(f"Total Belanja : Rp {total}")
```

# Penugasan 02. Hitung Faktorial

```python
# Input dari user
n = int(input("Masukkan angka: "))

# Inisialisasi variabel
faktorial = 1
i = 1

# Menghitung faktorial dengan while
while i <= n:
    faktorial *= i
    i += 1

# Output hasil faktorial
print("Faktorial dari", n, "adalah", faktorial)
```

# Penugasan 04. Cari Bilangan Prima dengan Break

```python
# Input dari user
n = int(input("Masukkan angka: "))

# Asumsi awal bahwa n adalah bilangan prima
is_prima = True

if n < 2:
    is_prima = False
else:
    for i in range(2, n):
        if n % i == 0:
            is_prima = False
            break

# Output hasil pengecekan
if is_prima:
    print(n, "adalah bilangan prima")
else:
    print(n, "bukan bilangan prima")
```

# Penugasan 08. Studi Kasus – Kasir Diskon

```python
total = 0
while True:
    barang = input("Masukkan nama barang (ketik 'selesai' untuk berhenti): ")
    if barang.lower() == "selesai":
        break
    harga = float(input(f"Masukkan harga {barang}: "))
    jumlah = int(input(f"Masukkan jumlah {barang}: "))
    subtotal = harga * jumlah
    total += subtotal

print("\n===== Struk Belanja =====")
print(f"Total Belanja : Rp {total}")

if total >= 200000:
    diskon = total * 0.1
    total_diskon = total - diskon
    print(f"Diskon 10%   : Rp {diskon}")
    print(f"Total Bayar  : Rp {total_diskon}")
else:
    print("Diskon 10%   : Rp 0")
    print(f"Total Bayar  : Rp {total}")
```
