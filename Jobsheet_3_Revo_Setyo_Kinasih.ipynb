#Praktikum 01 - Enkapsulasi pada kelas Bank Account

```python
class BankAccount:
    def __init__(self, owner, balance):
        # Atribut private
        self.__owner = owner
        self.__balance = balance

    def deposit(self, amount):
        """Method untuk menambahkan saldo."""
        if amount > 0:
            self.__balance += amount
            print(f"{amount} telah ditambahkan ke akun {self.__owner}.")
        else:
            print("Jumlah deposit harus lebih dari 0.")

    def withdraw(self, amount):
        """Method untuk menarik saldo."""
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"{amount} telah ditarik dari akun {self.__owner}.")
        else:
            print("Saldo tidak mencukupi.")

    def get_balance(self):
        """Method untuk mendapatkan informasi saldo terkini."""
        return self.__balance

# Contoh penggunaan
if __name__ == "__main__":
    alice_account = BankAccount(owner="Alice", balance=1000)
    alice_account.deposit(500)
    alice_account.withdraw(300)

    current_balance = alice_account.get_balance()
    print(f"Saldo terakhir di akun Alice: {current_balance}")
```

# Praktikum 02 - Enkapsulasi pada kelas Employee

```python
class Employee:
    def __init__(self, name, salary):
        # Atribut privat
        self.__name = name
        self.__salary = salary

    def update_salary(self, increase):
        """Method untuk menaikkan gaji dengan validasi."""
        if increase > 0:
            self.__salary += increase
            print(f"Gaji telah dinaikkan sebesar {increase}.")
        else:
            print("Nilai kenaikan harus lebih dari 0.")

    def set_salary(self, new_salary):
        """Method untuk mengubah gaji dengan validasi."""
        if new_salary >= 0:
            self.__salary = new_salary
            print(f"Gaji diatur ulang menjadi {new_salary}.")
        else:
            print("Gaji tidak dapat bernilai negatif.")

    def get_salary(self):
        return self.__salary

    def get_employee_info(self):
        """Method menampilkan informasi karyawan."""
        return f"Employee: {self.__name}, Gaji: {self.__salary}"

if __name__ == "__main__":
    employee1 = Employee("John Doe", 50000)
    print(employee1.get_employee_info())
    employee1.update_salary(5000)
    employee1.set_salary(60000)
    print(f"Informasi terbaru: {employee1.get_employee_info()}")
```

# Praktikum 03 - Kelas dan Objek Praktikum 03: Konstruktor dan Destruktor Sederhana

```python
class SimpleExample:
    def __init__(self, name):
        """Konstruktor: Dipanggil saat objek dibuat."""
        self.name = name
        print(f"Konstruktor: Objek '{self.name}' telah dibuat.")

    def __del__(self):
        """Destruktor: Dipanggil saat objek dihapus."""
        print(f"Destruktor: Objek '{self.name}' sedang dihapus.")

def main():
    print("Program dimulai.\n")
    obj = SimpleExample("Demo")
    print("Program sedang berjalan...\n")
    del obj # Menghapus objek secara eksplisit
    print("Objek telah dihapus secara eksplisit.\n")
    print("Program selesai.")

if __name__ == "__main__":
    main()
```

# Praktikum 04 - Konstruktor dan Destruktor program FileLogger

```python
class FileLogger:
    def __init__(self, filename):
        """Konstruktor: Membuka file log."""
        self.filename = filename
        try:
            self.file = open(filename, "a") # Mode append
            print(f"File '{filename}' berhasil dibuka untuk logging.")
        except Exception as e:
            print(f"Gagal membuka file '{filename}': {e}")

    def write_log(self, message):
        """Menulis pesan log ke dalam file."""
        self.file.write(message + "\n")
        self.file.flush() # Memastikan langsung tertulis ke disk
        print(f"Pesan log: '{message}' telah ditulis.")

    def __del__(self):
        """Destruktor: Menutup file log."""
        if hasattr(self, "file") and not self.file.closed:
            self.file.close()
            print(f"File '{self.filename}' telah ditutup.")

if __name__ == "__main__":
    logger = FileLogger("application.log")
    logger.write_log("Aplikasi dimulai.")
    logger.write_log("Operasi A selesai.")
    del logger
```

# Praktikum 05 -  Properti "self" pada kelas

```python
class Calculator:
    def __init__(self, initial_value=0):
        self.value = initial_value
        print(f"Kalkulator diinisialisasi dengan nilai: {self.value}")

    def add(self, number):
        self.value += number
        print(f"Setelah penambahan {number}, nilai sekarang adalah: {self.value}")

    def subtract(self, number):
        self.value -= number
        print(f"Setelah pengurangan {number}, nilai sekarang adalah: {self.value}")

    def show_value(self):
        print(f"Nilai saat ini adalah: {self.value}")

def main():
    calc1 = Calculator(initial_value=10)
    calc1.add(5)

    calc2 = Calculator() # Objek kedua independen
    calc2.add(20)

    calc1.show_value()
    calc2.show_value()

if __name__ == "__main__":
    main()
```

# Praktikum 06 - Menggunakan getter, setter, dan dekorator @property untuk mengelola akses atribut

```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    @property
    def name(self):
        """Getter untuk name."""
        return self.__name

    @name.setter
    def name(self, value):
        """Setter untuk name dengan validasi."""
        if not value:
            print("Nama tidak boleh kosong.")
        else:
            self.__name = value

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        """Setter untuk age dengan validasi[cite: 741]."""
        if value < 0:
            print("Umur tidak boleh negatif!")
        else:
            self.__age = value

if __name__ == "__main__":
    person = Person("Alice", 30)
    person.name = "Bob"
    person.age = -5 # Memicu validasi
    print(f"Nama: {person.name}, Umur: {person.age}")
```

# Buat Program

```python
class Student:
    def __init__(self, name, score):
        # Atribut privat
        self.__name = name
        self.__score = score
        self.__grade = ""
        # Update grade otomatis saat inisialisasi
        self.__update_grade()

    @property
    def name(self):
        """Getter untuk name."""
        return self.__name

    @name.setter
    def name(self, value):
        self.__name = value

    @property
    def score(self):
        """Getter untuk score."""
        return self.__score

    @score.setter
    def score(self, value):
        """Setter score dengan validasi 0-100."""
        if 0 <= value <= 100:
            self.__score = value
            self.__update_grade() # Update grade otomatis
        else:
            print("Error: Score harus berada di antara 0 hingga 100.")

    def __update_grade(self):
        """Method internal untuk menentukan grade berdasarkan skor."""
        if self.__score >= 90:
            self.__grade = "A"
        elif self.__score >= 80:
            self.__grade = "B"
        elif self.__score >= 70:
            self.__grade = "C"
        elif self.__score >= 60:
            self.__grade = "D"
        else:
            self.__grade = "E"

    def show_info(self):
        """Menampilkan informasi lengkap mahasiswa."""
        print(f"Nama Mahasiswa: {self.__name}")
        print(f"Nilai: {self.__score}")
        print(f"Grade: {self.__grade}")

    def __del__(self):
        """Destruktor."""
        print(f"Data mahasiswa {self.__name} telah dihapus dari sistem.")

# --- Pengujian Program (Sesuai Output yang Diminta) ---
if __name__ == "__main__":
    # 1. Membuat objek Siti dengan nilai 87
    mhs = Student("Siti", 87)
    mhs.show_info()

    # 2. Mengubah nilai menjadi 93
    print("\nNilai diubah...")
    mhs.score = 93
    mhs.show_info()

    # 3. Menghapus objek untuk memicu destruktor
    print()
    del mhs
```
