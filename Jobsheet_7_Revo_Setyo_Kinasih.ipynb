#Praktikum 1 : try. .except Dasar dalam Metode Kelas

```python
class DataStore:
    def __init__(self, nama_store: str = "Default Store"):
        self._data = {}  # Protected internal dictionary
        self.nama = nama_store

    def tambah_data(self, key, value):
        self._data[key] = value
        print(self._data)

    def get_data(self, key):
        """
        Mencoba mengakses self._data[key]
        menangani jika key tidak ada, mengembalikan None.
        """
        try:
            return self._data[key]
        except KeyError:
            print(f"Error: Kunci '{key}' tidak ditemukan di {self.nama}!")
            return None

# --- Kode Utama / Pengujian ---
if __name__ == "__main__":
    store = DataStore("Gudang Pusat")
    store.tambah_data("id_01", "Laptop ASUS")

    # Skenario Berhasil
    print("Hasil pencarian id_01:", store.get_data("id_01"))

    # Skenario Memicu KeyError
    print("Hasil pencarian id_99:", store.get_data("id_99"))
```

#Praktikum 2 : Menangani Beberapa Exception Spesifik

```python
class KalkulatorAngka:
    def __init__(self):
        pass

    def bagi(self, angka1, angka2):
        try:
            # Konversi input ke float
            num1 = float(angka1)
            num2 = float(angka2)

            # Lakukan pembagian
            hasil = num1 / num2
            return hasil

        except ValueError:
            print("Error: Input harus berupa angka valid (tidak bisa dikonversi ke float)!")
            return None
        except ZeroDivisionError:
            print("Error: Pembagian dengan nol tidak diperbolehkan secara matematika!")
            return None
        except Exception as e:
            print(f"Error tidak terduga terjadi: {e}")
            return None

# --- Kode Utama / Pengujian ---
if __name__ == "__main__":
    calc = KalkulatorAngka()

    print("--- Skenario Sukses ---")
    print("Hasil:", calc.bagi(50, 2))

    print("\n--- Memicu ZeroDivisionError ---")
    print("Hasil:", calc.bagi(50, 0))

    print("\n--- Memicu ValueError ---")
    print("Hasil:", calc.bagi("lima puluh", 2))
```

#Praktikum 3 : Menggunakan else dan finally dalam Metode

```python
import os

class FileProcessor:
    def __init__(self, nama_file: str):
        self.nama_file = nama_file
        self._file_handle = None  # Protected handle

    def baca_dan_proses(self):
        try:
            print(f"Membuka file: {self.nama_file}")
            # Buka file (bisa memicu FileNotFoundError)
            self._file_handle = open(self.nama_file, 'r')

            # Baca file
            konten = self._file_handle.read()
            print("Isi file sukses dibaca.")

        except FileNotFoundError:
            print(f"Error: File '{self.nama_file}' tidak ditemukan di direktori!")
        except Exception as e:
            print(f"Error saat memproses file: {e}")

        else:
            # Kode jika try berhasil (tanpa error)
            print("--- Menampilkan Konten ---")
            print(konten)
            print("--------------------------")

        finally:
            # Kode cleanup (selalu dijalankan)
            if self._file_handle and not self._file_handle.closed:
                self._file_handle.close()
                print("Cleanup: File handle berhasil ditutup dengan aman.")
            else:
                print("Cleanup: Tidak ada file handle terbuka yang perlu ditutup.")

# --- Kode Utama / Pengujian ---
if __name__ == "__main__":
    NAMA_FILE_SUKSES = "data_penting.txt"
    NAMA_FILE_GAGAL = "file_yang_tidak_ada.log"

    # Buat dummy file untuk skenario sukses
    with open(NAMA_FILE_SUKSES, "w") as dummy:
        dummy.write("Halo Revo! Ini adalah isi dari file dummy data_penting.txt.")

    print("\n--- Skenario 1: Proses Sukses ---")
    processor1 = FileProcessor(NAMA_FILE_SUKSES)
    processor1.baca_dan_proses()

    print("\n--- Skenario 2: Proses Gagal (File Tidak Ada) ---")
    processor2 = FileProcessor(NAMA_FILE_GAGAL)
    processor2.baca_dan_proses()

    # Bersihkan file dummy setelah pengujian selesai
    if os.path.exists(NAMA_FILE_SUKSES):
        os.remove(NAMA_FILE_SUKSES)
```

#Praktikum 4 : Mengakses Objek Exception (as e)

```python
class KalkulatorAngka:
    def __init__(self):
        print("Kalkulator Angka siap.")

    def bagi(self, angka1, angka2):
        """
        Membagi angka dengan angka2.
        Menangkap objek exception untuk menampilkan detail error.
        """
        print(f"\nMencoba membagi '{angka1}' dengan '{angka2}'...")
        try:
            num1 = float(angka1)
            num2 = float(angka2)
            if num2 == 0:
                # Bisa juga raise ZeroDivisionError di sini jika mau
                raise ZeroDivisionError("Pembagi tidak boleh nol secara eksplisit.")

            hasil = num1 / num2
            print(f" -> Hasil pembagian: {hasil}")
            return hasil

        # Menangkap ValueError dan menyimpannya di variabel 'e_val'
        except ValueError as e_val:
            print(" -> ERROR Konversi: Input tidak valid.")
            # Menampilkan representasi string dari objek exception e_val
            print(f"    Detail Error (ValueError): {e_val}")
            return None

        # Menangkap ZeroDivisionError dan menyimpannya di variabel 'e_zero'
        except ZeroDivisionError as e_zero:
            print(" -> ERROR Pembagian: Tidak dapat membagi dengan nol.")
            # Menampilkan representasi string dari objek exception e_zero
            print(f"    Detail Error (ZeroDivisionError): {e_zero}")
            return None

        # Menangkap exception lain dan menyimpannya di variabel 'e'
        except Exception as e:
            print(" -> ERROR Lainnya: Terjadi kesalahan tak terduga.")
            # Menampilkan representasi string dari objek exception e
            print(f"    Detail Error (Lainnya): {type(e).__name__} - {e}")
            return None


# --- Kode Utama ---
if __name__ == "__main__":
    calc = KalkulatorAngka()

    print("--- Tes Kalkulator ---")
    hasil1 = calc.bagi(100, 5)                    # Sukses
    hasil2 = calc.bagi(20, 0)                     # Memicu ZeroDivisionError
    hasil3 = calc.bagi("limapuluh", 10)           # Memicu ValueError
    hasil4 = calc.bagi(10, [])                    # Memicu TypeError (ditangkap except Exception)

    print("\n--- Ringkasan Hasil ---")
    print(f"Hasil 1 (100/5): {hasil1}")
    print(f"Hasil 2 (20/0): {hasil2}")
    print(f"Hasil 3 ('limapuluh'/10): {hasil3}")
    print(f"Hasil 4 (10/[]): {hasil4}")
```

#Praktikum 5 : Membangkitkan Exception (raise) dalam Metode

```python
import locale

# Skenario fallback locale tetap dipertahankan jika Anda butuh di device lain
try:
    locale.setlocale(locale.LC_ALL, 'id_ID.UTF-8')
except locale.Error:
    pass  # Tidak perlu print error di awal agar konsol tetap bersih

def format_rupiah(angka):
    """
    Memformat angka menjadi format Rupiah secara aman.
    Jika locale id_ID gagal, akan menggunakan format string manual yang universal.
    """
    try:
        return locale.currency(angka, grouping=True, symbol='Rp ')
    except (NameError, ValueError):
        # Menangkap ValueError jika locale 'C' bawaan sistem menolak memformat mata uang
        return f"Rp {angka:,.2f}".replace(",", ".")


class PengelolaAkun:
    def __init__(self, nomor_akun, saldo_awal=0):
        print(f"\nMencoba membuat akun {nomor_akun}...")
        if not isinstance(saldo_awal, (int, float)) or saldo_awal < 0:
            raise ValueError(f"Saldo awal ({saldo_awal}) tidak valid. Harus angka non-negatif.")

        self.nomor_akun = nomor_akun
        self.saldo = float(saldo_awal)
        print(f" -> Akun {self.nomor_akun} berhasil dibuat dengan saldo: {format_rupiah(self.saldo)}")

    def get_saldo(self):
        """Mengembalikan saldo saat ini."""
        return self.saldo

    def tarik_tunai(self, jumlah):
        """Menarik tunai dengan validasi. Me-raise exception jika gagal."""
        print(f"\n[{self.nomor_akun}] Mencoba tarik tunai: {format_rupiah(jumlah)}")

        if not isinstance(jumlah, (int, float)) or jumlah <= 0:
            raise ValueError(f"Jumlah penarikan ({jumlah}) tidak valid. Harus angka positif.")

        if jumlah > self.saldo:
            raise RuntimeError(f"Saldo tidak mencukupi! Saldo: {format_rupiah(self.saldo)}, Diminta: {format_rupiah(jumlah)}")

        self.saldo -= jumlah
        print(f" -> Penarikan {format_rupiah(jumlah)} berhasil.")
        print(f" -> Saldo sekarang: {format_rupiah(self.saldo)}")
        return True


# --- Kode Utama ---
if __name__ == "__main__":
    # Inisialisasi variabel dengan None agar tidak memicu NameError jika skenario sebelumnya gagal
    akun_sukses = None

    # Skenario 1: Gagal membuat akun karena saldo awal tidak valid
    print("\n--- Skenario 1: Saldo Awal Tidak Valid ---")
    try:
        akun_gagal = PengelolaAkun("ACC001", -50000)
    except ValueError as e:
        print(f"   ERROR DITANGKAP saat pembuatan akun: {e}")

    # Skenario 2: Membuat akun valid, lalu coba transaksi
    print("\n--- Skenario 2: Akun Valid dan Transaksi ---")
    try:
        akun_sukses = PengelolaAkun("ACC002", 100000)

        # Transaksi 1: Sukses
        akun_sukses.tarik_tunai(30000)

        # Transaksi 2: Gagal karena jumlah tidak valid (ValueError)
        print("\n Mencoba tarik jumlah negatif...")
        akun_sukses.tarik_tunai(-10000)

    except ValueError as e:
        print(f"   ERROR DITANGKAP (ValueError): {e}")
    except RuntimeError as e:
        print(f"   ERROR DITANGKAP (RuntimeError): {e}")

    # Skenario 3: Melanjutkan akun valid, gagal karena saldo tidak cukup
    print("\n--- Skenario 3: Saldo Tidak Cukup ---")
    try:
        # Cek apakah objek akun_sukses berhasil dibuat di Skenario 2
        if akun_sukses is not None:
            print(f"   Saldo akun ACC002 saat ini: {format_rupiah(akun_sukses.get_saldo())}")
            print("   Mencoba tarik jumlah > saldo...")
            akun_sukses.tarik_tunai(80000)  # Saldo hanya sisa 70000
        else:
            print("   Skenario 3 dilewati karena akun_sukses gagal dibuat pada skenario sebelumnya.")

    except ValueError as e:
        print(f"   ERROR DITANGKAP (ValueError): {e}")
    except RuntimeError as e:
        print(f"   ERROR DITANGKAP (RuntimeError): {e}")  # Ini yang diharapkan tertangkap

    # Cetak saldo akhir dengan aman
    if akun_sukses is not None:
        print(f"\nSaldo akhir akun ACC002: {format_rupiah(akun_sukses.get_saldo())}")
```

#Praktikum 6 : Kombinasi Menangkap dan Membangkitkan Exception

```python
# 1. Definisikan Custom Exception (mewarisi dari Exception)
class KonfigurasiError(Exception):
    """Exception khusus untuk menandakan error terkait konfigurasi aplikasi."""

    def __init__(self, message, penyebab_asli=None):
        super().__init__(message)
        self.penyebab_asli = penyebab_asli # Menyimpan error asli jika ada

    def __str__(self):
        if self.penyebab_asli:
            return f"{super().__str__()} (Disebabkan oleh: {type(self.penyebab_asli).__name__})"
        return super().__str__()


# Kelas yang menggunakan exception handling gabungan
class PengelolaSetting:
    def __init__(self, file_config):
        self.config = {}
        self.file_config = file_config
        # Panggil metode internal yang berpotensi raise KonfigurasiError
        try:
            self._load_config()
            print(f"-> PengelolaSetting berhasil dibuat untuk '{self.file_config}'.")
        except KonfigurasiError as e:
            print(f"-> GAGAL membuat PengelolaSetting: {e}")
            # Bisa jadi kita ingin raise lagi agar pembuat objek tahu ada masalah serius
            # raise e
            # Atau set state default / kosong
            self.config = {}
            print("   -> Menggunakan konfigurasi default/kosong.")

    def _load_config(self):
        """Metode internal untuk load config. Bisa raise KonfigurasiError."""
        print(f"\n [{self.file_config}] Mencoba load konfigurasi...")
        try:
            # --- Potensi FileNotFoundError ---
            print(f"   Membuka file '{self.file_config}'...")
            # Simulasi: anggap hanya file 'config.txt' yang valid untuk dibaca
            if self.file_config != 'config.txt':
                # Jika file tidak ada, kita raise FileNotFoundError secara manual
                # (dalam aplikasi nyata, ini akan otomatis dari open())
                raise FileNotFoundError(f"File '{self.file_config}' tidak ada.")

            # --- Potensi Error Isi File (KeyError, ValueError) ---
            print(f"   Membaca dan validasi isi '{self.file_config}'...")
            # Simulasi data raw (bisa dari json.load(f), dll)

            # CATATAN SIMULASI:
            # Untuk Skenario 1: (Akan loncat ke except FileNotFoundError terlebih dahulu)
            # Untuk Skenario 2: Aktifkan data_raw dengan port salah tipe ("xyz")
            # Untuk Skenario 3: Aktifkan data_raw yang benar

            # Silakan sesuaikan komentar baris data_raw di bawah ini saat menguji skenario:

            # Pilihan A (Untuk Skenario 2): Port salah tipe
            data_raw = {"host": "191.168.1.1", "port": "xyz"}

            # Pilihan B (Untuk Skenario 3): Data benar (uncomment baris di bawah jika ingin tes sukses)
            # data_raw = {"host": "192.168.1.1", "port": "8080"}

            # Pilihan C: Key 'port' hilang
            # data_raw = {"host": "192.168.1.1"}

            # Cek key 'port'
            port_str = data_raw['port'] # Bisa raise KeyError

            # Cek tipe 'port'
            port_int = int(port_str) # Bisa raise ValueError

            # Jika lolos, simpan
            self.config['port'] = port_int
            self.config['host'] = data_raw.get('host', 'localhost')
            print("   -> Konfigurasi berhasil di-load dan divalidasi.")

        # --- Blok Except ---
        except FileNotFoundError as e_fnf:
            print("   ERROR Bawaan: File tidak ditemukan.")
            # Re-raise sebagai KonfigurasiError
            raise KonfigurasiError(f"File konfigurasi penting '{self.file_config}' hilang.", penyebab_asli=e_fnf) from e_fnf

        except KeyError as e_key:
            print("   ERROR Bawaan: Kunci konfigurasi hilang.")
            # Re-raise sebagai KonfigurasiError
            raise KonfigurasiError(f"Kunci '{e_key}' wajib ada dalam file konfigurasi.", penyebab_asli=e_key) from e_key

        except ValueError as e_val:
            print("   ERROR Bawaan: Nilai konfigurasi tidak valid.")
            # Re-raise sebagai KonfigurasiError
            raise KonfigurasiError(f"Nilai untuk 'port' ('{port_str}') harus integer.", penyebab_asli=e_val) from e_val

        except Exception as e_lain:
            print(f"   ERROR Bawaan Tak Terduga: {type(e_lain).__name__}")
            # Re-raise sebagai KonfigurasiError umum
            raise KonfigurasiError("Terjadi error tak terduga saat memuat konfigurasi.", penyebab_asli=e_lain) from e_lain

    def get_setting(self, key, default=None):
        """Mendapatkan nilai setting."""
        # print(f"Mendapatkan setting '{key}'...")
        return self.config.get(key, default)


# --- Kode Utama ---
if __name__ == "__main__":
    # Skenario 1: File tidak ada -> Menghasilkan KonfigurasiError
    print("\n--- Skenario 1: File Salah ---")
    # Error sudah ditangani di dalam __init__ berdasarkan contoh ini
    settings1 = PengelolaSetting("setting.cfg")
    # Jika error di-raise ulang oleh init, kode di bawah tidak jalan
    print(f"Port dari settings1: {settings1.get_setting('port', 'Belum terload')}")

    # Skenario 2: File ada, tapi data port salah -> Menghasilkan KonfigurasiError
    print("\n--- Skenario 2: Data Port Salah ---")
    # (Anggap 'config.txt' ada tapi isinya menyebabkan ValueError di _load_config)
    settings2 = PengelolaSetting("config.txt")
    print(f"Port dari settings2: {settings2.get_setting('port', 'Belum terload')}")

    # Skenario 3: (Jika simulasi di _load_config diubah jadi benar)
    print("\n--- Skenario 3: Sukses ---")
    print("*(Catatan: Untuk mencoba Skenario 3, pastikan variabel data_raw yang benar di dalam fungsi _load_config diaktifkan)*")
    settings3 = PengelolaSetting("config.txt")
    print(f"Port dari settings3: {settings3.get_setting('port', 'Belum terload')}")
    print(f"Host dari settings3: {settings3.get_setting('host', 'Belum terload')}")
```

#Penugasan 3 : Program dan Kelas Diagram

```python
class SistemPendaftaranKaryawan:
    def __init__(self):
        self.database_karyawan = {}

    def daftar_karyawan(self, id_karyawan, nama, umur, email):
        try:
            # --- Validasi Aturan (Poin a s.d d) ---

            # a. ID Karyawan harus berupa string alfanumerik & tepat 5 karakter
            if not isinstance(id_karyawan, str) or not id_karyawan.isalnum() or len(id_karyawan) != 5:
                raise ValueError("ID Karyawan harus berupa alfanumerik dan panjangnya tepat 5 karakter!")

            # b. ID Karyawan bersifat unik (belum pernah terdaftar)
            if id_karyawan in self.database_karyawan:
                raise KeyError(f"ID Karyawan '{id_karyawan}' sudah terdaftar dalam sistem!")

            # c. Umur harus integer dan berada dalam rentang 18 s.d 60 tahun
            if not isinstance(umur, int):
                raise TypeError("Umur harus berupa bilangan bulat (Integer)!")
            if not (18 <= umur <= 60):
                raise ValueError("Umur karyawan harus berada di antara 18 hingga 60 tahun!")

            # d. Email harus valid (mengandung karakter '@' dan '.')
            if not isinstance(email, str) or '@' not in email or '.' not in email:
                raise ValueError("Format email tidak valid! Email wajib mengandung karakter '@' dan '.'")

        # --- Penanganan Exception Spesifik (Poin e) ---
        except TypeError as te:
            print(f"[PENDAFTARAN GAGAL]: Terjadi kesalahan tipe data -> {te}")
            return False
        except ValueError as ve:
            print(f"[PENDAFTARAN GAGAL]: Aturan validasi dilanggar -> {ve}")
            return False
        except KeyError as ke:
            print(f"[PENDAFTARAN GAGAL]: Duplikasi data ditemukan -> {ke}")
            return False
        except Exception as e:
            print(f"[PENDAFTARAN GAGAL]: Terjadi error yang tidak terduga -> {e}")
            return False

        # --- Blok Optional Kondisional (Poin f) ---
        else:
            # Dijalankan hanya jika blok try bersih dari exception
            self.database_karyawan[id_karyawan] = {
                "nama": nama,
                "umur": umur,
                "email": email
            }
            print(f"[PENDAFTARAN SUKSES]: Karyawan baru bernama '{nama}' ({id_karyawan}) berhasil disimpan!")
            return True

        finally:
            # Selalu dijalankan di akhir proses pendaftaran karyawan apa pun statusnya
            print(f"[LOG SISTEM]: Sesi pendaftaran untuk ID Karyawan '{id_karyawan}' telah selesai diproses.\n")


# --- Unit Testing / Pengujian Sistem ---
if __name__ == "__main__":
    sistem = SistemPendaftaranKaryawan()

    print("=== PENGUJIAN 1: Pendaftaran Valid (Sukses) ===")
    sistem.daftar_karyawan("KRW01", "Revo Setyo", 20, "revo@polines.ac.id")

    print("=== PENGUJIAN 2: Memicu KeyError (ID Sudah Ada) ===")
    sistem.daftar_karyawan("KRW01", "Budi Utomo", 25, "budi@gmail.com")

    print("=== PENGUJIAN 3: Memicu ValueError (Panjang ID Salah) ===")
    sistem.daftar_karyawan("KRW99999", "Cici", 30, "cici@gmail.com")

    print("=== PENGUJIAN 4: Memicu TypeError (Umur Bukan Integer) ===")
    sistem.daftar_karyawan("KRW02", "Dedi", "dua puluh lima", "dedi@gmail.com")

    print("=== PENGUJIAN 5: Memicu ValueError (Umur di Bawah Batas) ===")
    sistem.daftar_karyawan("KRW02", "Dedi", 17, "dedi@gmail.com")

    print("=== PENGUJIAN 6: Memicu ValueError (Format Email Salah) ===")
    sistem.daftar_karyawan("KRW03", "Eka", 45, "ekagmailcom")

    # Menampilkan total data karyawan yang sukses masuk ke database sistem
    print("=== ISI DATABASE AKHIR ===")
    print(sistem.database_karyawan)
```
